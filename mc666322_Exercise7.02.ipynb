```python
%matplotlib inline

# 导入必要的库
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 设置 Seaborn 默认样式
sns.set()

# 读取数据
data = pd.read_csv("northern_surface_temperature.csv", index_col=['Year'])
```

```python
# 转置数据，使行变为月份，列变为年份
data = data.transpose()
```

```python
# 创建自定义发散调色板：低值蓝色（240°），高值红色（15°），饱和度为 99
heat_colormap = sns.diverging_palette(240, 15, s=99, as_cmap=True)

# 绘制热图，每 5 年取一列，center=0 确保零值对应中性色
plt.figure(dpi=200)
sns.heatmap(data.iloc[:, ::5], cmap=heat_colormap, center=0)
plt.title("Temperature Changes from 1880 to 2015 (base period 1951-1980)")

# 保存图像
plt.savefig('temperature_change.png', dpi=300, bbox_inches='tight')

# 显示图形
plt.show()
```
