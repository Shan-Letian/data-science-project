```python
%matplotlib inline

# 导入必要的库
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 设置 Seaborn 默认样式
sns.set()
```

```python
# 读取数据
data = pd.read_csv("flight_details.csv")
```

```python
# 将数据透视成矩阵：行是月份，列是年份，值是乘客数
# 先按月份顺序重新排列
month_order = ['January', 'February', 'March', 'April', 'May', 'June',
               'July', 'August', 'September', 'October', 'November', 'December']
data['Months'] = pd.Categorical(data['Months'], categories=month_order, ordered=True)

# 透视
pivot_data = data.pivot(index='Months', columns='Years', values='Passengers')
```

```python
# 绘制热图
plt.figure(figsize=(10, 6), dpi=150)
sns.heatmap(pivot_data, cmap='GnBu', annot=False, fmt='d',
            cbar_kws={'label': 'Passengers'})

# 添加标题和坐标轴标签
plt.title('Flight Passengers from 1949 to 1960')
plt.xlabel('Years')
plt.ylabel('Months')

# 显示图形
plt.tight_layout()
plt.show()
```
