```python
%matplotlib inline

# 导入所需库
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 读取数据（请根据实际路径调整文件名或路径）
mydata = pd.read_csv("iq_scores.csv")
```

```python
# 将每列数据转换为列表
group_a = mydata[mydata.columns[0]].tolist()
group_b = mydata[mydata.columns[1]].tolist()
group_c = mydata[mydata.columns[2]].tolist()
group_d = mydata[mydata.columns[3]].tolist()
```

```python
# 打印各组数据，确认转换成功（可选）
print(group_a)
print(group_b)
print(group_c)
print(group_d)
```

```python
# 构建长格式 DataFrame
data = pd.DataFrame({
    'Groups': ['Group A'] * len(group_a)
              + ['Group B'] * len(group_b)
              + ['Group C'] * len(group_c)
              + ['Group D'] * len(group_d),
    'IQ score': group_a + group_b + group_c + group_d
})
```

```python
# 查看列名（可选）
print(data.columns)
```

```python
# 设置 Seaborn 风格和上下文
sns.set_style('whitegrid')
sns.set_context('talk')

# 创建图形
plt.figure(dpi=150)

# 绘制箱线图
sns.boxplot(x='Groups', y='IQ score', data=data)

# 去除多余边框：左、右、上
sns.despine(left=True, right=True, top=True)

# 添加标题
plt.title('IQ scores for different test groups')

# 显示图形
plt.show()
```
