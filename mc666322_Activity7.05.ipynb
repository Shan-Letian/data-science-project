```python
# 01: Read the dataset
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("anage_data.csv")
```

```python
# 02: Filter Mammalia samples with body mass < 200,000 g
filtered_df = df[
    (df["Class"] == "Mammalia") &
    (df["Body mass (g)"] < 200000)
].dropna(subset=["Body mass (g)", "Maximum longevity (yrs)"])

# Optional: check the filtered data
print(filtered_df[["Body mass (g)", "Maximum longevity (yrs)"]].head())
print("Number of samples:", len(filtered_df))
```

```python
# 03: Create the regression plot
sns.set_theme()

plt.figure(figsize=(8, 5))
sns.regplot(
    data=filtered_df,
    x="Body mass (g)",
    y="Maximum longevity (yrs)",
    scatter_kws={"alpha": 0.6}
)

plt.title("Linear Regression for Animal Attribute Relations")
plt.xlabel("Body mass (g)")
plt.ylabel("Maximum longevity (yrs)")
plt.tight_layout()
plt.show()
```
