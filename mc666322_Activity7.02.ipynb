```python
%matplotlib inline

# 导入必要的库
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 设置 Seaborn 默认样式
sns.set()
```

```python
# 读取数据
data = pd.read_csv("movie_scores.csv")
```

```python
# 转换为长格式
df = data.melt(id_vars='MovieTitle',
               value_vars=['AudienceScore', 'Tomatometer'],
               var_name='Type', value_name='Scores')

# 把列名改成更易读的显示形式
df['Type'] = df['Type'].replace({'Tomatometer': 'Tomatometer',
                                 'AudienceScore': 'Audience Score'})
```

```python
# 创建图形
plt.figure(figsize=(10, 6), dpi=150)

# 绘制分组条形图
sns.barplot(x='MovieTitle', y='Scores', hue='Type', data=df)

# 添加标题和标签
plt.title('Movies Scores comparison')
plt.xlabel('Movies')
plt.ylabel('Scores')

# 旋转 x 轴标签，避免重叠
plt.xticks(rotation=15, ha='right')

# 显示图例
plt.legend(title='Type')

# 调整布局
plt.tight_layout()
plt.show()
```
