```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import squarify
```

```python
# 读取数据集
mydata = pd.read_csv("water_usage.csv", index_col=0)
```

```python
# 创建标签，例如：Leak (12%)
labels = mydata['Usage'] + ' (' + mydata['Percentage'].astype('str') + '%)'

# 创建图形
plt.figure(dpi=200)

# 创建树状图
squarify.plot(
    sizes=mydata['Percentage'],
    label=labels,
    color=sns.light_palette('green', mydata.shape[0])
)

# 关闭坐标轴
plt.axis('off')

# 添加标题
plt.title('Water usage')

# 显示图形
plt.show()
```
